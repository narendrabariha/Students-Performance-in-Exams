```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import os

file_path = os.path.expanduser("~/Downloads/StudentsPerformance.xls")

print(os.path.exists(file_path))
print(file_path)
```

```python
file_path = r"C:\Users\Hp/Downloads/StudentsPerformance.xls"

df = pd.read_csv(file_path)
```

```python
print("\nFirst 5 Rows:")
print(df.head())

print("\nLast 5 Rows:")
print(df.tail())

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nDataset Information:")
print(df.info())

print("\nStatistical Summary:")
print(df.describe())
```

```python
print("\nMissing Values:")
print(df.isnull().sum())

print("\nTotal Missing Values:")
print(df.isnull().sum().sum())
```

```python

print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

```python
for column in df.columns:
    print(f"\n{column}:")
    print(df[column].unique())
```

```python
print("\nGender Distribution:")
print(df["gender"].value_counts())

print("\nRace/Ethnicity Distribution:")
print(df["race/ethnicity"].value_counts())

print("\nParental Education:")
print(df["parental level of education"].value_counts())

print("\nLunch Distribution:")
print(df["lunch"].value_counts())

print("\nTest Preparation:")
print(df["test preparation course"].value_counts())
```

```python
df["total_score"] = (
    df["math score"]
    + df["reading score"]
    + df["writing score"]
)

df["average_score"] = df["total_score"] / 3

print("\nDataset with Total and Average Score:")
print(df.head())
```

```python
print("\nAverage Math Score:")
print(df["math score"].mean())

print("\nAverage Reading Score:")
print(df["reading score"].mean())

print("\nAverage Writing Score:")
print(df["writing score"].mean())

print("\nAverage Total Score:")
print(df["total_score"].mean())

print("\nAverage Overall Score:")
print(df["average_score"].mean())
```

```python
print("\nHighest Math Score:")
print(df["math score"].max())

print("\nLowest Math Score:")
print(df["math score"].min())

print("\nHighest Overall Score:")
print(df["average_score"].max())

print("\nLowest Overall Score:")
print(df["average_score"].min())
```

```python
gender_performance = df.groupby("gender")[
    ["math score", "reading score", "writing score"]
].mean()

print("\nGender-wise Performance:")
print(gender_performance)
```

```python
education_performance = df.groupby(
    "parental level of education"
)[
    ["math score", "reading score", "writing score"]
].mean().sort_values("math score", ascending=False)

print("\nPerformance by Parental Education:")
print(education_performance)
```

```python
preparation_performance = df.groupby(
    "test preparation course"
)[
    ["math score", "reading score", "writing score"]
].mean()

print("\nTest Preparation vs Performance:")
print(preparation_performance)
```

```python
lunch_performance = df.groupby("lunch")[
    ["math score", "reading score", "writing score"]
].mean()

print("\nLunch Type vs Performance:")
print(lunch_performance)
```

```python
race_performance = df.groupby(
    "race/ethnicity"
)[
    ["math score", "reading score", "writing score"]
].mean()

print("\nRace/Ethnicity vs Performance:")
print(race_performance)
```

```python
correlation = df[
    ["math score", "reading score", "writing score"]
].corr()

print("\nCorrelation Matrix:")
print(correlation)
```

```python
sns.set_theme(style="whitegrid")
```

```python
plt.figure(figsize=(10, 4))

sns.histplot(
    df["average_score"],
    bins=20,
    kde=True
)

plt.title("Distribution of Average Student Scores")
plt.xlabel("Average Score")
plt.ylabel("Number of Students")

plt.show()
```

```python
# Chart 2: Gender Performance
# --------------------------------------------

plt.figure(figsize=(10, 3))

sns.barplot(
    data=df,
    x="gender",
    y="average_score"
)

plt.title("Average Score by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Score")

plt.show()
```

```python
# Chart 3: Test Preparation
# --------------------------------------------

plt.figure(figsize=(10, 3))

sns.barplot(
    data=df,
    x="test preparation course",
    y="average_score"
)

plt.title("Test Preparation Course vs Performance")
plt.xlabel("Test Preparation")
plt.ylabel("Average Score")

plt.show()


# --------------------------------------------
# Chart 4: Parental Education
# --------------------------------------------

plt.figure(figsize=(12, 3))

sns.barplot(
    data=df,
    x="parental level of education",
    y="average_score"
)

plt.title("Parental Education vs Student Performance")
plt.xlabel("Parental Education")
plt.ylabel("Average Score")

plt.xticks(rotation=45)

plt.show()


# --------------------------------------------
# Chart 5: Lunch vs Performance
# --------------------------------------------

plt.figure(figsize=(8, 3))

sns.barplot(
    data=df,
    x="lunch",
    y="average_score"
)

plt.title("Lunch Type vs Student Performance")
plt.xlabel("Lunch Type")
plt.ylabel("Average Score")

plt.show()
```

```python
# Chart 6: Math vs Reading
# --------------------------------------------

plt.figure(figsize=(8, 3))

sns.scatterplot(
    data=df,
    x="math score",
    y="reading score"
)

plt.title("Math Score vs Reading Score")
plt.xlabel("Math Score")
plt.ylabel("Reading Score")

plt.show()


# --------------------------------------------
# Chart 7: Correlation Heatmap
# --------------------------------------------

plt.figure(figsize=(8, 3))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Between Student Scores")

plt.show()
```

```python
# 18. Top 10 Students
# ============================================

top_students = df.sort_values(
    "average_score",
    ascending=False
).head(10)

print("\nTop 10 Students:")
print(
    top_students[
        [
            "gender",
            "race/ethnicity",
            "math score",
            "reading score",
            "writing score",
            "average_score"
        ]
    ]
)


# ============================================
# 19. Students Needing Improvement
# ============================================

low_performers = df[
    df["average_score"] < 50
]

print("\nStudents with Average Score Below 50:")
print(low_performers.shape[0])


# ============================================
# 20. Pass/Fail Analysis
# ============================================

df["result"] = np.where(
    df["average_score"] >= 50,
    "Pass",
    "Fail"
)

print("\nPass/Fail Distribution:")
print(df["result"].value_counts())


# ============================================
# 21. Pass Percentage
# ============================================

pass_percentage = (
    (df["result"] == "Pass").mean() * 100
)

print(
    f"\nOverall Pass Percentage: "
    f"{pass_percentage:.2f}%"
)


# ============================================
# 22. Save Cleaned Dataset
# ============================================

df.to_csv(
    "StudentsPerformance_Analyzed.csv",
    index=False
)

print("\nAnalyzed dataset saved successfully!")
```
